```python
import pandas as pd
```

```python
data = pd.read_csv('../FUT_PYTHON_STRATEGY/data_total/dados_betfair.csv', sep=';')

# Seleicionar as colunas relevantes
data = data[['Home', 'Away', 'Odd_H_Back', 'Odd_H_Lay', 'Odd_A_Back', 'Odd_A_Lay', 'Odd_D_Back', 'Odd_D_Lay', 'Odd_CS_0x1_Lay',
'Goals_H_HT', 'Goals_A_HT', 'Goals_H_FT', 'Goals_A_FT', 'Min_Goals_A', 'Min_Goals_H']]

# Crie uma nova coluna Profit / Loss para o lay 0 x 1 | descontar 4,5 de comissão nas apostas vencedoras no else 1
data['Profit_Lay_0x1'] = round(data.apply(lambda row: (1 - row['Odd_CS_0x1_Lay']) if row['Goals_A_FT'] == 1 and row['Goals_H_FT'] == 0 else 1 * 0.955, axis=1), 2)

# Salvar o DataFrame atualizado em um novo arquivo CSV
data.to_csv('../FUT_PYTHON_STRATEGY/data_total/dados_betfair_atualizado.csv', index=False, sep=';')

data.head()
```
